```python
sns.scatterplot(x=data['weekdaysort'],y=data['monthdaysort'])
plt.title('sorte des jours et des mois')
plt.show()
```

```python
#compter le nombre de cafe par jour de la semaine
graphique=data.groupby('Time_of_Day')['coffee_name'].count()
graphique.plot(kind='bar')
plt.title('nombre de cafe par jour')
plt.xlabel('Time_of_Day')
plt.ylabel('nombre de cafe')
plt.show()
```
